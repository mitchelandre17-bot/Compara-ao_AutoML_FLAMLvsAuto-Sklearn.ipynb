6. Importando as bibliotecas e carregando os dados

```python
import pandas as pd
import numpy as np
import time

from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# carregar dataset Iris
iris = load_iris()

# separar caracteristicas e variavel alvo
x = iris.data
y = iris.target

# mostrar informaçoes do dataset
print("Quantidade de registros:", len(x))
print("Características:", iris.feature_names)
print("Classes:", iris.target_names)
```

7. dividindo o treino e o teste

```python
# dividir os dados em treino e teste
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Dados de treinamento:", x_train.shape)
print("Dados de teste:", x_test.shape)
```

8. Configurar e treinar

```python
from flaml import AutoML

# criar o modelo AutoML
flaml_model = AutoML()

# registrar o tempo inicial
inicio_flaml = time.time()

# configuraçoes do FLAML
flaml_model.fit(
    x_train,
    y_train,

    task="classification",

    time_budget=60,

    metric="accuracy",

    eval_method="cv",

    seed=42,

    n_jobs=1
)

# registrar o tempo final
fim_flaml = time.time()

# calcular tempo de execuçao
tempo_flaml = fim_flaml - inicio_flaml

print("Treinamento FLAML concluído!")
print("Tempo de execução:", round(tempo_flaml, 2), "segundos")
```

9. Avaliando o FLAML

```python
# realizar previsoes
previsoes_flaml = flaml_model.predict(x_test)

# calcular acuracia
acuracia_flaml = accuracy_score(
    y_test,
    previsoes_flaml
)

print("ACURÁCIA FLAML:", round(acuracia_flaml, 4))

# relatorio de classificaçao
print("\nRelatório de classificação:")

print(
    classification_report(
        y_test,
        previsoes_flaml,
        target_names=iris.target_names
    )
)

# matriz de confusao
print("Matriz de confusão:")

print(
    confusion_matrix(
        y_test,
        previsoes_flaml
    )
)

# melhor algoritmo encontrado
print("\nMelhor estimador:")
print(flaml_model.best_estimator)

# melhores configuraçoes
print("\nMelhor configuração:")
print(flaml_model.best_config)
```

10. Treinar o Auto-Sklearn

```python
%%writefile /content/treinar_autosklearn.py

import time

from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from autosklearn.classification import AutoSklearnClassifier


# carregar dataset Iris
iris = load_iris()

x = iris.data
y = iris.target


# dividir treino e teste
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)


# criar modelo Auto-Sklearn
autosklearn_model = AutoSklearnClassifier(

    time_left_for_this_task=60,

    per_run_time_limit=20,

    seed=42,

    n_jobs=1,

    memory_limit=3072,

    ensemble_size=20,

    resampling_strategy="cv",

    resampling_strategy_arguments={
        "folds": 5
    }
)


# iniciar contagem do tempo
inicio = time.time()


# treinar modelo
autosklearn_model.fit(
    x_train,
    y_train,
    dataset_name="iris"
)


# finalizar contagem
fim = time.time()

tempo = fim - inicio


# realizar previsoes
previsoes = autosklearn_model.predict(x_test)


# calcular acuracia
acuracia = accuracy_score(
    y_test,
    previsoes
)


# mostrar resultados
print("\n========== AUTO-SKLEARN ==========")

print("\nAcurácia:", round(acuracia, 4))

print("\nTempo de execução:", round(tempo, 2), "segundos")


# relatorio de classificaçao
print("\nRelatório de classificação:")

print(
    classification_report(
        y_test,
        previsoes,
        target_names=iris.target_names
    )
)


# matriz de confusao
print("\nMatriz de confusão:")

print(
    confusion_matrix(
        y_test,
        previsoes
    )
)


# mostrar estatisticas da busca
print("\nEstatísticas da busca:")

print(
    autosklearn_model.sprint_statistics()
)


# mostrar modelos encontrados
print("\nModelos e pesos do ensemble:")

print(
    autosklearn_model.show_models()
)


# mostrar ranking dos modelos
print("\nRanking dos modelos:")

print(
    autosklearn_model.leaderboard()
)
```
